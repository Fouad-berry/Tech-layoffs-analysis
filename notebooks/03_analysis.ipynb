# 03 — Business Analysis & Insights

Key questions answered with charts ready for the Looker dashboard.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

sns.set_theme(style='whitegrid')
df = pd.read_csv('../data/processed/layoffs_clean.csv', parse_dates=['date'])
print(df.shape)
```

## Q1. Which industries shed the most jobs?

```python
by_industry = df.groupby('industry')['layoffs_count'].sum().sort_values()
by_industry.plot(kind='barh', figsize=(10, 5), color='tomato')
plt.title('Total Layoffs by Industry')
plt.xlabel('Total Layoffs')
plt.tight_layout()
plt.show()
```

## Q2. Layoffs vs. open roles over time

```python
quarterly = df.groupby('year_quarter').agg(
    total_layoffs=('layoffs_count', 'sum'),
    avg_open_roles=('open_roles', 'mean')
).reset_index()

fig, ax1 = plt.subplots(figsize=(12, 5))
ax2 = ax1.twinx()
ax1.bar(quarterly['year_quarter'], quarterly['total_layoffs'], color='tomato', alpha=0.7, label='Total Layoffs')
ax2.plot(quarterly['year_quarter'], quarterly['avg_open_roles'], color='steelblue', marker='o', label='Avg Open Roles')
ax1.set_xlabel('Quarter')
ax1.set_ylabel('Total Layoffs', color='tomato')
ax2.set_ylabel('Avg Open Roles', color='steelblue')
plt.title('Layoffs vs. Open Roles by Quarter')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Q3. Does higher AI adoption correlate with more layoffs?

```python
fig = px.scatter(
    df.sample(2000, random_state=42),
    x='ai_adoption_level',
    y='layoff_percentage',
    color='industry',
    size='layoffs_count',
    hover_data=['company_name', 'country'],
    title='AI Adoption vs. Layoff %',
    opacity=0.6,
)
fig.show()
```

## Q4. Market condition impact on employee sentiment

```python
fig = px.box(
    df,
    x='market_condition',
    y='employee_sentiment',
    color='market_condition',
    title='Employee Sentiment by Market Condition'
)
fig.show()
```

## Q5. AI risk tier breakdown by industry

```python
pivot = df.groupby(['industry', 'ai_risk_tier']).size().unstack(fill_value=0)
pivot.plot(kind='bar', stacked=True, figsize=(12, 6), colormap='RdYlGn_r')
plt.title('AI Risk Tier by Industry')
plt.xlabel('Industry')
plt.ylabel('Number of Events')
plt.legend(title='AI Risk Tier')
plt.tight_layout()
plt.show()
```

# 01 — Exploratory Data Analysis

Quick overview of the dataset: shape, distributions, nulls, correlations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
pd.set_option('display.max_columns', None)

df = pd.read_csv('../data/raw/tech_layoffs_hiring_trends.csv')
print(f'Shape: {df.shape}')
df.head()
```

```python
# Null values
nulls = df.isnull().sum()
print(nulls[nulls > 0])
print('\nDtypes:')
print(df.dtypes)
```

```python
# Numeric distributions
df.describe().T
```

```python
# Categorical value counts
cats = ['industry', 'country', 'company_size', 'hiring_trend', 'market_condition', 'reason_for_layoffs']
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
for ax, col in zip(axes.flat, cats):
    counts = df[col].value_counts()
    counts.plot(kind='bar', ax=ax, color='steelblue')
    ax.set_title(col)
    ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap for numeric columns
num_cols = df.select_dtypes('number').columns
plt.figure(figsize=(14, 10))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
# Layoffs distribution
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
df['layoffs_count'].hist(bins=40, ax=ax1, color='tomato')
ax1.set_title('Layoffs Count Distribution')
df['layoff_percentage'].hist(bins=40, ax=ax2, color='steelblue')
ax2.set_title('Layoff % Distribution')
plt.tight_layout()
plt.show()
```
